```python
from json import load, dump
a = list(load(open('../../py65536py16_.json')))
m = load(open('../../py256.json'))
merge = []
for _,i in enumerate(a):
    while 1:
        J = 0
        for j in range(len(i)-1, 0, -1):
            if j==1 or i[:j]in a:
                J = J or j
                if j==len(i)-1 or i[j:]in a:
                    merge.append((i[:j],i[j:]))
                    break
        else:
            i = i[J:]
            a[_] = i
            continue
        break
m['model']['merges'] = merge
m['model']['vocab'] = {a:i for i,a in enumerate(list(m['model']['vocab'])[:64]+a)}
dump(m, open('../../py65536py16.json','w'), ensure_ascii=0, separators=',:')
```
