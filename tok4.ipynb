```python
from heapq import heapify, heappop, heappushpop
from multiprocessing.pool import ThreadPool
from mmap import mmap
def reduce(counts, vocab=256):
    heap = [(len(i)*count,i)for i,count in counts.items()]
    heapify(heap)
    push = None #delayed push for pushpop
    while len(counts) > vocab:
        score,i = heappushpop(heap,push) if push else heappop(heap)
        push = None
        count = counts.get(i,0)
        if score == len(i)*count:
            del counts[i]
            if len(i) > 2:
                i = i[:-1]
                count += counts.get(i,0)
                counts[i] = count
                push = len(i)*count,i
    return counts
def count(dat, gram=4, vocab=256, buf=2560):
    counts = {}
    for i in range(len(dat)-gram+1):
        i = dat[i:i+gram]
        counts[i] = counts.get(i, 0) + 1
        if len(counts) >= buf:
            reduce(counts, vocab)
    return counts
def merge(*counts):
    counter = counts[0]
    for counts in counts[1:]:
        for i,count in counts.items():
            counter[i] = counter.get(i,0) + count
    return counter
file = open('../../wikitext2.txt')
dat = memoryview(mmap(file.fileno(), file.tell()))
with ThreadPool(12) as pool:
    print(reduce(merge(*pool.map(count, [dat[i:i+len(dat)//12]for i in range(0,len(dat),len(dat)//12)]))))

# Convert this to high performance 
```
